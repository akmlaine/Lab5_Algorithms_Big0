```python
import random

# Student-specific inputs
LAST_NAME = "Malitao"
STUDENT_ID = "TUPM-26-1135"
FAVORITE_ARTIST = "ADIE"

digits = "".join(char for char in STUDENT_ID if char.isdigit())

SEED_DIGIT = int(digits[-1])
ID_SUM = sum(int(digit) for digit in digits)
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(char) - ord("A") + 1 for char in FAVORITE_ARTIST)

SEED = ID_SUM + ARTIST_SUM + NAME_LENGTH + ARTIST_LENGTH
rng = random.Random(SEED)

# Generate original incident records
original_records = sorted(rng.sample(range(100, 1000), 12))

# Rotate the records
rotation = SEED_DIGIT % len(original_records)

if rotation == 0:
    rotated_records = original_records.copy()
else:
    rotated_records = (
        original_records[-rotation:] + original_records[:-rotation]
    )

# Select target using student-specific values
target_index = (ID_SUM + ARTIST_SUM) % len(original_records)
target = original_records[target_index]


def rotated_binary_search(records, target):
    left = 0
    right = len(records) - 1
    comparisons = 0

    while left <= right:
        mid = (left + right) // 2
        comparisons += 1

        if records[mid] == target:
            return mid, comparisons

        comparisons += 1
        if records[left] <= records[mid]:
            comparisons += 1
            if records[left] <= target < records[mid]:
                right = mid - 1
            else:
                left = mid + 1
        else:
            comparisons += 1
            if records[mid] < target <= records[right]:
                left = mid + 1
            else:
                right = mid - 1

    return -1, comparisons


def linear_search(records, target):
    comparisons = 0

    for index, value in enumerate(records):
        comparisons += 1
        if value == target:
            return index, comparisons

    return -1, comparisons


binary_position, binary_comparisons = rotated_binary_search(
    rotated_records, target
)

linear_position, linear_comparisons = linear_search(
    rotated_records, target
)

print("EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH")
print("Original records:", original_records)
print("Rotated records:", rotated_records)
print("Target record:", target)

if binary_position != -1:
    print("Binary search position (1-based):", binary_position + 1)
else:
    print("Binary search: Target not found")

print("Binary search comparisons:", binary_comparisons)

if linear_position != -1:
    print("Linear search position (1-based):", linear_position + 1)
else:
    print("Linear search: Target not found")

print("Linear search comparisons:", linear_comparisons)

print("Binary search time complexity: O(log n)")
print("Linear search time complexity: O(n)")
print("Auxiliary space complexity: O(1)")
```

```python
# Student-specific inputs
LAST_NAME = "Malitao"
STUDENT_ID = "TUPM-26-1135"
FAVORITE_ARTIST = "ADIE"

digits = "".join(char for char in STUDENT_ID if char.isdigit())

SEED_DIGIT = int(digits[-1])
ID_SUM = sum(int(digit) for digit in digits)
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

# Generate ascending enrollment records
department_a = sorted(
    ID_SUM + SEED_DIGIT + i * (NAME_LENGTH + 1)
    for i in range(8)
)

department_b = sorted(
    ARTIST_LENGTH + NAME_LENGTH + i * (SEED_DIGIT + 2)
    for i in range(7)
)


def merge_records(first, second):
    i = 0
    j = 0
    merged = []
    comparisons = 0

    while i < len(first) and j < len(second):
        comparisons += 1

        if first[i] <= second[j]:
            merged.append(first[i])
            i += 1
        else:
            merged.append(second[j])
            j += 1

    while i < len(first):
        merged.append(first[i])
        i += 1

    while j < len(second):
        merged.append(second[j])
        j += 1

    return merged, comparisons


merged_records, comparisons = merge_records(
    department_a, department_b
)

print("EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS")
print("Department A records:", department_a)
print("Department B records:", department_b)
print("Merged records:", merged_records)
print("Comparisons during merging:", comparisons)

print("Total records before merging:",
      len(department_a) + len(department_b))
print("Total records after merging:", len(merged_records))
print("Records are in ascending order:",
      merged_records == sorted(merged_records))
print("Record count verified:",
      len(merged_records) == len(department_a) + len(department_b))

print("Time complexity: O(n + m)")
print("Auxiliary space complexity: O(n + m)")
```

```python
import random
import time

# Student-specific inputs
LAST_NAME = "Malitao"
STUDENT_ID = "TUPM-26-1135"
FAVORITE_ARTIST = "ADIE"

digits = "".join(char for char in STUDENT_ID if char.isdigit())

SEED_DIGIT = int(digits[-1])
ID_SUM = sum(int(digit) for digit in digits)
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(char) - ord("A") + 1 for char in FAVORITE_ARTIST)

SEED = ID_SUM + ARTIST_SUM + NAME_LENGTH + ARTIST_LENGTH

rng = random.Random(SEED + 100)
generated_orders = rng.sample(range(1000, 9999), 10)

# Three input conditions
input_cases = {
    "Sorted": sorted(generated_orders),
    "Reverse sorted": sorted(generated_orders, reverse=True),
    "Random": generated_orders.copy()
}

rng.shuffle(input_cases["Random"])


def bubble_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0
    n = len(arr)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 2
                swapped = True

        if not swapped:
            break

    return arr, comparisons, movements


def selection_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0
    n = len(arr)

    for i in range(n - 1):
        min_index = i

        for j in range(i + 1, n):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            movements += 2

    return arr, comparisons, movements


def insertion_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        arr[j + 1] = key
        movements += 1

    return arr, comparisons, movements


algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

print("EXERCISE 3: E-COMMERCE ORDER PROCESSING")
print("Generated orders:", generated_orders)

for case_name, records in input_cases.items():
    print("\nInput condition:", case_name)
    print("Input records:", records)

    for algorithm_name, algorithm in algorithms.items():
        start_time = time.perf_counter()

        sorted_records, comparisons, movements = algorithm(records)

        elapsed_time = time.perf_counter() - start_time

        print("\nAlgorithm:", algorithm_name)
        print("Sorted records:", sorted_records)
        print("Comparisons:", comparisons)
        print("Movements:", movements)
        print("Execution time:", f"{elapsed_time:.8f} seconds")
        print("Sorting verified:", sorted_records == sorted(records))

print("\nTIME COMPLEXITY")
print("Bubble Sort: Best O(n), Average O(n^2), Worst O(n^2)")
print("Selection Sort: Best/Average/Worst O(n^2)")
print("Insertion Sort: Best O(n), Average O(n^2), Worst O(n^2)")
```
